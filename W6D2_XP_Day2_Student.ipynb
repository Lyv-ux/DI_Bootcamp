## 🌟 Exercise 2: LLM Architecture and Application Scenarios
For each, describe (a) core architectural differences, (b) a real application, (c) why it fits.

### BERT
- **Core Architectural Differences:** Bidirectional encoder-only architecture. Pre-trained using Masked Language Modeling (MLM) to predict masked tokens based on both left and right context, and historically, Next Sentence Prediction (NSP) to understand sentence relationships.
- **Application:** Named Entity Recognition (NER), Question Answering (QA), Sentiment Analysis.
- **Why:** Its bidirectional nature allows it to deeply understand the context of words within a sentence, making it excellent for tasks requiring a rich contextual representation of the input text for classification or extraction.

### GPT
- **Core Architectural Differences:** Decoder-only architecture. Operates as an autoregressive causal language model, meaning it generates text sequentially, predicting the next token based on all preceding tokens.
- **Application:** Chatbots, content generation (articles, creative writing), code generation, text summarization from scratch.
- **Why:** Its autoregressive nature and ability to generate coherent, contextually relevant text makes it ideal for tasks that require producing long, flowing sequences of text based on a given prompt or prior context.

### T5
- **Core Architectural Differences:** Encoder-decoder architecture. Unifies all NLP tasks into a "text-to-text" framework, where every problem is converted into generating an output text sequence given an input text sequence.
- **Application:** Machine Translation, Text Summarization, Question Answering (generative), Paraphrasing.
- **Why:** The encoder-decoder structure is inherently suited for sequence-to-sequence tasks. The encoder processes the input text, and the decoder generates the corresponding output text, making it highly versatile for transforming one text sequence into another for a wide range of NLP problems.

## 🌟 Exercise 3: Benefits and Ethics of Pre-training
- Benefits (explain each in your words):
  1) Improved generalization: Pre-trained models learn vast patterns and relationships from large datasets, allowing them to perform well even on tasks or data they haven't explicitly seen during fine-tuning.
  2) Less labeled data: Because pre-trained models already have a foundational understanding of language, they require significantly less task-specific labeled data for fine-tuning compared to training models from scratch.
  3) Faster fine-tuning: The pre-training phase provides a strong starting point, meaning the model converges much quicker during fine-tuning on a specific downstream task, saving computational resources and time.
  4) Transfer learning: Knowledge gained from pre-training on a general, often massive, dataset (like Wikipedia or Common Crawl) can be effectively transferred and adapted to various specific NLP tasks, even those with limited data.
  5) Robustness: Pre-training on diverse and extensive datasets makes models more robust to variations, noise, and ambiguities in real-world language inputs, leading to more stable performance.

- Ethical concerns:
  - **Bias:** Models can inadvertently learn and perpetuate societal biases present in their training data, leading to unfair or discriminatory outputs (e.g., gender, racial, or cultural bias).
  - **Misinformation:** Generative models can produce highly convincing but false information, potentially contributing to the spread of disinformation and propaganda.
  - **Misuse:** Powerful LLMs can be misused for malicious purposes, such as generating spam, creating deepfakes, automating social engineering attacks, or developing harmful content.
  - **Privacy:** Training data may contain sensitive personal information that could inadvertently be reproduced or inferred by the model, raising privacy concerns.

- Mitigations:
  - **Data curation:** Carefully selecting, cleaning, and balancing training datasets to reduce the presence of harmful biases and sensitive information.
  - **Differential privacy:** Implementing techniques during training to obscure individual data points, preventing the model from memorizing or reproducing specific private information.
  - **Safety filters:** Developing post-processing filters or guardrails to detect and prevent models from generating harmful, toxic, or biased content.
  - **Reinforcement Learning from Human Feedback (RLHF):** Fine-tuning models with human preferences to guide them toward generating helpful, harmless, and honest outputs.
  - **Audits and transparency:** Regularly auditing models for fairness, bias, and performance, and maintaining transparency about their capabilities, limitations, and data sources.

## 🌟 Exercise 4: Transformer Architecture Deep Dive
### Self-Attention & Multi-Head Attention
- **Q/K/V flow, softmax weighting, and value mixing:** In self-attention, each input token is transformed into three vectors: Query (Q), Key (K), and Value (V). The Query vector of a token is multiplied by the Key vectors of all other tokens (including itself) to compute attention scores. These scores are then scaled and passed through a softmax function to get attention weights, indicating how much focus each token should place on others. Finally, these attention weights are multiplied by the Value vectors of all tokens and summed up, resulting in a weighted sum that represents the contextually enriched representation of the original token.
- **Why multiple heads help (subspace projections, diverse relations):** Multiple attention heads allow the model to simultaneously focus on different aspects or relationships within the input sequence. Each head learns to project the Q, K, and V vectors into different lower-dimensional subspaces, effectively allowing it to attend to different parts of the input or capture diverse types of relationships (e.g., syntactic dependencies, semantic similarities) independently. The outputs from these multiple heads are then concatenated and linearly transformed, providing a richer and more comprehensive contextual understanding.
- **Example sentence (different from lesson):** "The swift fox jumped over the lazy dog."
  - **Relation 1 (Head 1):** A head might capture verb-subject relationships. For "jumped", it might heavily attend to "fox".
  - **Relation 2 (Head 2):** Another head might capture adjective-noun relationships. For "fox", it might attend to "swift", and for "dog", it might attend to "lazy".

### Pre-training Objectives
- **MLM vs CLM (compare objectives and typical use):**
  - **Masked Language Modeling (MLM):** The objective is to predict randomly masked tokens in a sequence, given the unmasked tokens from both left and right contexts. **Typical use:** Encoder-only models like BERT, used for tasks requiring a deep understanding of the entire input, such as classification or token-level prediction.
  - **Causal Language Modeling (CLM):** The objective is to predict the next token in a sequence, given only the preceding tokens (unidirectional context). **Typical use:** Decoder-only models like GPT, used for generative tasks where the model needs to produce sequential, coherent text.
- **When to prefer MLM vs CLM:**
  - **Prefer MLM** for understanding-heavy tasks where a full contextual grasp of the input is needed (e.g., sentiment analysis, question answering, summarization *by extraction*). It's suitable when you want the model to process an entire input sequence to make a decision or prediction about that sequence.
  - **Prefer CLM** for generation-heavy tasks where the model needs to produce new text sequentially (e.g., chatbots, creative writing, code generation, summarization *by generation*). It's ideal when the output is a direct continuation or creation based on a given prompt.
- **NSP (why early BERT used it and why many modern models avoid it):** Next Sentence Prediction (NSP) was used in early BERT pre-training to help the model understand the relationship between two sentences. The task was to predict whether two sentences followed each other in the original text. BERT used it because it was thought to be crucial for downstream tasks like Question Answering and Natural Language Inference, which require understanding inter-sentence coherence. Many modern models (e.g., RoBERTa, ELECTRA) avoid NSP because subsequent research showed it contributed less to overall performance than MLM, and sometimes even hindered it. Focusing solely on MLM, or using more robust alternatives for sentence-level understanding, proved more effective.

### Transformer Model Selection
- **Sentiment on reviews:** Encoder-only (e.g., BERT, RoBERTa).
  - **Justification:** Sentiment analysis is a classification task that requires a deep, bidirectional understanding of the entire input text (the review) to classify its overall sentiment. Encoder-only models excel at creating rich contextual representations for such tasks.
- **Conversational chatbot (creative responses):** Decoder-only (e.g., GPT, Llama).
  - **Justification:** Chatbots that generate creative and open-ended responses need to produce text sequentially, token by token, based on the conversation history. Decoder-only models are inherently designed for autoregressive text generation and excel at creating coherent and contextually relevant continuations.
- **Technical document translation (EN→ES):** Encoder-Decoder (e.g., T5, BART).
  - **Justification:** Machine translation is a classic sequence-to-sequence task. An encoder-decoder architecture is ideal because the encoder can fully understand the source language sentence (English), and then the decoder can generate the target language sentence (Spanish) based on that understanding, transforming one text sequence into another.

### Positional Encoding
- **Purpose (why absolute/relative position info matters):** Transformers process input tokens in parallel, meaning they inherently lose information about the order or position of tokens within a sequence. Positional encodings are added to the token embeddings to inject this crucial positional information. Without it, the model would treat a sequence like "Dog bites man" the same as "Man bites dog," which have very different meanings. Positional encoding allows the model to understand the absolute position of each token and, through attention, to infer relative positions and relationships between tokens based on their distances.
- **Example failure without positions:** Consider the sentences:
  1. "**The boy** kicked **the ball**."
  2. "**The ball** kicked **the boy**."
  Without positional encoding, the transformer's self-attention mechanism would only see the words 'The', 'boy', 'kicked', 'the', 'ball' as a bag of words, regardless of their order. It wouldn't be able to distinguish between the subject and the object, leading to a complete failure in understanding the meaning and likely producing incorrect representations for both sentences.

## 🌟 Exercise 6: Softmax Temperature: The Randomness Regulator
### 1) Temperature Scenarios
- **T=0.2:** Very low temperature. The model becomes highly deterministic, picking the most probable next token with a very high likelihood. Output will be very focused, conservative, and less creative, often repetitive or generic.
- **T=1.5:** High temperature. The model assigns more uniform probabilities across all possible next tokens, increasing the likelihood of selecting less probable, more 'surprising' tokens. Output becomes more creative, diverse, and potentially incoherent or nonsensical.
- **T=1.0:** Default temperature (no scaling). The model's raw logits are used directly for probability distribution. This represents a baseline level of randomness, offering a balance between coherence and creativity, reflecting the inherent probabilities learned during training.

### 2) Application Design
- **Bedtime stories (creativity vs coherence):** Use a moderate to high temperature (e.g., T=0.8 to T=1.2) with a decoding strategy like `top_k` or `top_p` sampling. **Why:** Bedtime stories need to be engaging and creative to hold attention, but also coherent enough to follow a narrative. A slightly higher temperature encourages imaginative word choices without completely sacrificing readability. `top_k`/`top_p` can help prune truly outlandish options.
- **Financial report summaries (accuracy/reliability):** Use a very low temperature (e.g., T=0.1 to T=0.5) or even greedy decoding (effectively T->0). **Why:** Financial reports demand absolute accuracy, factual correctness, and reliability. There is no room for creativity or hallucination. A low temperature ensures the model selects the most statistically probable and factual tokens, minimizing deviation from the source material and reducing the risk of generating false information.

### 3) Temperature & Bias
- **How does temperature influence bias surfacing or dampening? Give a realistic example.**
  Temperature can significantly influence how bias surfaces or is dampened. At **low temperatures**, the model tends to be more deterministic and will more reliably produce responses that align with the dominant patterns (including biases) present in its training data. If the training data contains stereotypes, a low temperature will make the model more likely to consistently reproduce those stereotypes because they represent the most statistically probable next tokens. For example, if a model has learned a strong bias associating 'engineer' with 'male', at a low temperature, a prompt like "The engineer went to..." is highly likely to be completed with a male pronoun ("his office").

  Conversely, at **high temperatures**, the model's output becomes more diverse and less predictable. This can sometimes *dampen* obvious biases by making the model select less probable (and potentially less biased) tokens, leading to more varied and less stereotypical responses. However, it can also *surface* less common or more subtle biases that might not be dominant at lower temperatures, or even lead to unexpected and potentially offensive combinations due to increased randomness. For example, at a high temperature, the model might occasionally use female pronouns for an engineer or generate a wildly imaginative, but still subtly biased, scenario that would not appear at a lower temperature. The increased randomness can make the emergence of bias less predictable, sometimes masking it, but also potentially revealing different facets of the underlying biased distribution.
