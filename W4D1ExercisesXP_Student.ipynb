```python

# This piece of code is already prefilled, run it to execute it and see the results.
# It demonstrates standard classification metrics for binary loan default prediction.

from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score, average_precision_score, confusion_matrix, classification_report

# Please replace these placeholders with your true labels and predicted probabilities.
y_true = [0,1,0,1,0,0,1,0,1,0]            # placeholder labels
y_pred_proba = [0.05,0.80,0.10,0.65,0.20,0.15,0.70,0.30,0.85,0.25]  # placeholder probabilities

# You should set a decision threshold that reflects the precision–recall trade-off for your business case.
threshold = 0.5
y_pred = [1 if p >= threshold else 0 for p in y_pred_proba]

print("Accuracy:", round(accuracy_score(y_true, y_pred), 4))
print("Precision:", round(precision_score(y_true, y_pred, zero_division=0), 4))
print("Recall:", round(recall_score(y_true, y_pred, zero_division=0), 4))
print("F1-score:", round(f1_score(y_true, y_pred, zero_division=0), 4))
print("ROC-AUC:", round(roc_auc_score(y_true, y_pred_proba), 4))
print("PR-AUC (Average Precision):", round(average_precision_score(y_true, y_pred_proba), 4))
print("\nConfusion matrix:\n", confusion_matrix(y_true, y_pred))
print("\nClassification report:\n", classification_report(y_true, y_pred, zero_division=0))
```

The F1-score provides a balance between precision and recall. ROC-AUC (Receiver Operating Characteristic - Area Under Curve) assesses the model's ability to distinguish between classes across all possible thresholds, while PR-AUC (Precision-Recall Area Under Curve) is particularly informative for imbalanced datasets, focusing on the performance of the positive class. The decision threshold will be chosen by analyzing the precision-recall curve and selecting a point that balances the cost of false positives and false negatives according to the specific business objectives and risk tolerance of the financial institution.

**1. Predicting Stock Prices: predict future prices**

**Suitable Machine Learning Type:** Supervised Learning (Regression).

**Justification:** Predicting stock prices involves forecasting a continuous numerical value based on historical data. Supervised learning is appropriate because we have labeled data (past stock prices and related features) that can be used to train a model to map inputs to target outputs. Regression is specifically chosen because the output is a continuous number (the future stock price) rather than a discrete category.

**Input Data:** The input data would consist of historical stock prices (open, close, high, low), trading volumes, company financial reports, economic indicators, news sentiment, and technical analysis indicators (e.g., moving averages, RSI) over various time periods.

**Output:** The output would be a continuous numerical value representing the predicted future price of a stock (e.g., the closing price for the next day or week).

**Learning Signal/Objective:** The learning signal is the difference between the model's predicted stock price and the actual future stock price. The objective is to minimize this prediction error (e.g., Root Mean Squared Error - RMSE or Mean Absolute Error - MAE) and maximize the accuracy of the forecast.

**2. Organizing a Library of Books: group books into genres or categories based on similarities.**

**Suitable Machine Learning Type:** Unsupervised Learning (Clustering).

**Justification:** Organizing a library of books into genres or categories based on similarities, without pre-defined labels for these categories, is a classic clustering problem. Unsupervised learning is suitable because the goal is to discover inherent patterns and groupings within the data without any explicit target variable. Clustering algorithms can identify natural clusters of books that share similar characteristics.

**Input Data:** The input data would include textual features such as book titles, authors, summaries, full text content (extracted keywords, topics, sentiment), publication dates, and potentially metadata like ISBNs or genre tags from external sources if available for similarity comparison.

**Output:** The output would be a set of clusters, where each book is assigned to one or more clusters representing its genre or category. The clusters themselves are derived by the algorithm based on the similarities it finds.

**Learning Signal/Objective:** There is no explicit learning signal in the traditional sense, as there are no 'correct' labels. The objective is to maximize intra-cluster similarity (books within the same cluster are very similar) and minimize inter-cluster similarity (books in different clusters are dissimilar), often measured by metrics like silhouette score or Davies-Bouldin index.

**3. Program a robot to navigate and find the shortest path in a maze.**

**Suitable Machine Learning Type:** Reinforcement Learning.

**Justification:** Programming a robot to navigate and find the shortest path in a maze is an optimal control problem where an agent (the robot) learns to make a sequence of decisions (movements) in an environment (the maze) to achieve a goal (reaching the exit) by maximizing a cumulative reward (minimizing path length). Reinforcement learning is ideal because the robot learns through trial and error, receiving feedback (rewards or penalties) for its actions.

**Input Data:** The input data to the robot would be its current state in the maze, which could include its current position (coordinates), sensory input (e.g., distance to walls, presence of obstacles in different directions), and potentially a map of the maze if available.

**Output:** The output would be an action or a sequence of actions (e.g., move forward, turn left, turn right) that the robot takes at each step to navigate the maze.

**Learning Signal/Objective:** The learning signal is the reward or penalty the robot receives after each action (e.g., a small negative reward for each step taken, a large negative reward for hitting a wall, and a large positive reward for reaching the exit). The objective is to learn a policy (a strategy for choosing actions based on states) that maximizes the cumulative reward over time, effectively finding the shortest path to the exit.

Supervised Learning Model Evaluation (e.g., Loan Default Prediction - Classification)

For a supervised classification model, such as one predicting loan defaults, the evaluation strategy would involve a robust validation plan to ensure reliability and generalization. The dataset would first be split into training, validation, and test sets, typically with an 80/10/10 ratio respectively, ensuring stratified sampling to maintain the class distribution of the target variable across all sets. During model development and hyperparameter tuning, k-fold cross-validation will be applied on the training data, allowing the model to be trained and validated on different subsets multiple times to reduce variance in performance estimation. The primary metrics to report include Accuracy, Precision, Recall, F1-score, ROC-AUC, and PR-AUC (Average Precision). ROC curves and Precision-Recall curves will be plotted to visualize classifier performance across various thresholds, and a decision threshold will be chosen that balances the business costs of false positives (e.g., denying a good applicant) and false negatives (e.g., approving a defaulter). A key evaluation challenge for supervised models is data leakage, where information from the test set inadvertently influences model training or hyperparameter selection, leading to overly optimistic performance estimates.

Unsupervised Learning Model Evaluation (e.g., Organizing a Library of Books - Clustering)

For an unsupervised clustering model, like grouping books into genres, the evaluation focuses on assessing the quality and meaningfulness of the discovered clusters, as there are no true labels for comparison. We would measure cluster quality primarily using intrinsic metrics that evaluate the inherent structure of the data based on the clustering result. The Silhouette Score is a key metric, quantifying how similar an object is to its own cluster compared to other clusters, with higher values indicating better-defined clusters. The Davies-Bouldin Index would also be reported, where a lower value indicates better clustering as it represents a better separation between clusters. Diagnostic plots like the silhouette plot would be used to visualize cluster separation and identify potential issues, and a dendrogram could help understand the hierarchical structure of the clusters. The Elbow Method would be employed to determine the optimal number of clusters by plotting the within-cluster sum of squares (WCSS) against the number of clusters and looking for the 'elbow point' where the rate of decrease significantly slows. A key evaluation challenge for unsupervised models is the lack of ground truth, making objective quantitative evaluation difficult and often requiring domain expertise to interpret cluster meaning.

Reinforcement Learning Model Evaluation (e.g., Robot Navigation in a Maze)

For a reinforcement learning model, such as a robot navigating a maze, success is measured by the agent's ability to learn an optimal policy that maximizes cumulative reward over time. We would track the cumulative reward per episode, which is the sum of rewards received by the agent from the start to the end of a task, as a primary indicator of performance; consistently increasing cumulative reward suggests effective learning. Convergence is assessed by observing whether the cumulative reward stabilizes over many episodes, indicating that the agent has learned a stable policy and is no longer significantly improving. The balance between exploration (trying new actions to discover better policies) and exploitation (using known good actions to maximize immediate reward) is critical; this can be evaluated by observing the agent's behavior in different phases of learning and measuring the frequency of exploratory actions. Visualizations of the agent's path in the maze for successful and unsuccessful runs would also provide qualitative insights. A key evaluation challenge for reinforcement learning models is sample efficiency, meaning they often require a vast number of interactions with the environment to learn an optimal policy, which can be computationally expensive or impractical in real-world scenarios
