🌟 Exercice 2 : Statistiques descriptives

```python
import numpy as np
from scipy import stats

data = [12, 15, 13, 12, 18, 20, 22, 21]

moyenne = stats.tmean(data)
mediane = np.median(data)

variance = stats.tvar(data)  
ecart_type = stats.tstd(data)

print(f"La moyenne est: {moyenne}")
print(f"La médiane est : {mediane}") #On utilise NUmpy car SciPy ne sait pas faire ça
print(f"La variance est: {variance:.4f}")
print(f"L'écart-type : {ecart_type:.4f}")

```

🌟 Exercice 3 : Comprendre les distributions

```python
from scipy import stats
import numpy as np
import matplotlib.pyplot as plt

# Gaussian Distribution
mu, sigma = 16.5, 4.13  # mean and standard deviation
gaussian_dist = stats.norm(mu, sigma)
print("PDF at x=4.13:", gaussian_dist.pdf(4.13))

# Plotting Gaussian Distribution
x_gaussian = np.linspace(1, 100, 100) #Définir un ensemble de points
pdf_gaussian = gaussian_dist.pdf(x_gaussian) #
plt.figure(figsize=(6, 4))
plt.plot(x_gaussian, pdf_gaussian)
plt.title('Gaussian Distribution\n$\mu=0$, $\sigma=0.1$')
plt.xlabel('x')
plt.ylabel('PDF')
plt.show()
```

🌟 Exercice 4 : Application du test T

```python
from scipy import stats


data1 = stats.norm.rvs(loc=5, scale=10, size=100)
data2 =  stats.norm.rvs(loc=5, scale=10, size=100)

t_statistic, p_value = stats.ttest_ind(data1, data2)

print(f"Statistique T : {t_statistic:.4f}")
print(f"P-value : {p_value}")

# Interprétation standard (alpha = 0.05)
alpha = 0.05
if p_value < alpha:
    print("Conclusion : La différence entre les deux moyennes est statistiquement significative (on rejette l'hypothèse nulle H0).")
else:
    print("Conclusion : Il n'y a pas de différence statistiquement significative entre les deux moyennes (on ne peut pas rejeter H0).")
```

🌟 Exercice 5 : Analyse de régression linéaire

```python
import numpy as np
from scipy import stats

# Données
house_sizes = np.array([50, 70, 80, 100, 120])
house_prices = np.array([150000, 200000, 210000, 250000, 280000])

# Calcul de la régression linéaire
slope, intercept, r_value, p_value, std_err = stats.linregress(house_sizes, house_prices)

print(f"Pente (Slope) : {slope:.2f}")
print(f"Intersection (Intercept) : {intercept:.2f}")

# Prédiction pour 90 m²
taille_cible = 90
prix_predit = slope * taille_cible + intercept
print(f"Prix prédit pour une maison de {taille_cible} m² : {prix_predit:.2f} unités monétaires")
```

🌟 Exercice 6 : Comprendre l’ANOVA

```python
from scipy import stats

# Données
fertilizer_1 = [5, 6, 7, 6, 5]
fertilizer_2 = [7, 8, 7, 9, 8]
fertilizer_3 = [4, 5, 4, 3, 4]

# Test ANOVA à un facteur
f_stat, p_value = stats.f_oneway(fertilizer_1, fertilizer_2, fertilizer_3)

print(f"Valeur F (F-statistic) : {f_stat:.4f}")
print(f"Valeur P (P-value) : {p_value:.4e}")
```

Réponses aux questions :

1.Valeurs F et P : La valeur $F$ est de 23.6522 et la valeur $P$ est d'environ 7.23e-05 (soit $0.0000723$).

2.Effets significatifs ? Oui. Comme la valeur $P$ ($0.0000723$) est très inférieure au seuil standard de $0.05$, on conclut que les engrais ont des effets significativement différents sur la croissance des plantes. Au moins un des engrais produit un résultat différent des autres.

3.Si la valeur P était supérieure à 0.05 : Si $P > 0.05$, on n'aurait pas pu rejeter l'hypothèse nulle ($H_0$). Cela signifierait qu'il n'y a pas de preuve statistique que les engrais ont des effets différents ; les variations de croissance observées seraient alors considérées comme le simple fruit du hasard.

🌟 Exercice 7 : Distributions de probabilité (optionnelle)

```python
from scipy import stats

# Paramètres : n = nombre de lancers (10), p = probabilité de succès (0.5 pour "face")
n = 10
p = 0.5
k = 5  # Nombre exact de succès voulus

# Calcul de la probabilité exacte
prob_exact_5 = stats.binom.pmf(k, n, p)

print(f"La probabilité d'obtenir exactement 5 fois 'face' sur 10 lancers est de : {prob_exact_5:.4f} (soit {prob_exact_5*100:.2f}%)")
```

🌟 Exercice 8 : Coefficients de corrélation (optionnel)

```python
import pandas as pd
from scipy import stats

# Données
data = pd.DataFrame({
    'age': [23, 25, 30, 35, 40], 
    'income': [35000, 40000, 50000, 60000, 70000]
})

# Corrélation de Pearson (linéaire)
pearson_corr, p_val_p = stats.pearsonr(data['age'], data['income'])

# Corrélation de Spearman (monotone / basée sur les rangs)
spearman_corr, p_val_s = stats.spearmanr(data['age'], data['income'])

print(f"Coefficient de Pearson : {pearson_corr:.4f}")
print(f"Coefficient de Spearman : {spearman_corr:.4f}")
```
