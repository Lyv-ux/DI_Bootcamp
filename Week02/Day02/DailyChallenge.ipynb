```python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler

# 1. Importation du jeu de données et suppression de la colonne des id 
df = pd.read_csv('datascience_salaries.csv', index_col = 0)


# 2. Normalisation Min-Max de la colonne 'salary'
scaler = MinMaxScaler()
df["salary_normalized"] = scaler.fit_transform(df[["salary"]])

# 3. Préparation et réduction de dimensionnalité (PCA)
# Sélection des colonnes numériques pertinentes pour l'analyse géométrique
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns
df_numeric = df[numeric_cols].dropna()

# Standardisation rapide avant PCA (essentiel pour la variance)
df_numeric_scaled = (df_numeric - df_numeric.mean()) / df_numeric.std()

# Application de la PCA pour réduire à 2 dimensions principales
pca = PCA(n_components=2)
pca_result = pca.fit_transform(df_numeric_scaled)

df["pca_1"] = pca_result[:, 0]
df["pca_2"] = pca_result[:, 1]

# 4. Agrégation des données par 'experience_level'
salary_stats = (
    df.groupby("experience_level")["salary"]
    .agg(["mean", "median"])
    .reset_index()
)


salary_stats["experience_level"] = salary_stats["experience_level"]


# --- Affichage des résultats ---
print("=== APPERÇU DES DONNÉES NORMALISÉES ET PCA ===")
print(df[["experience_level", "salary", "salary_normalized", "pca_1", "pca_2"]].head())

print("\n=== STATISTIQUES DES SALAIRES PAR NIVEAU D'EXPÉRIENCE ===")
print(salary_stats.to_string(index=False))
```
