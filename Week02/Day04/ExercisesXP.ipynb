🌟 Exercice 1 : Comprendre la visualisation des données
Importance de la visualisation dans l'analyse des données
La visualisation des données est essentielle car le cerveau humain traite les images beaucoup plus rapidement que le texte ou les tableaux de chiffres bruts. Elle permet de :

Identifier des tendances et des motifs (patterns) : Repérer instantanément des structures ou des cycles qui seraient invisibles dans un tableau Excel.

Détecter les anomalies : Isoler rapidement les valeurs aberrantes (outliers) ou les erreurs dans les données.

Faciliter la prise de décision : Traduire des données complexes en insights clairs et exploitables pour les décideurs.

Storytelling : Raconter une histoire convaincante et partager des conclusions de manière universelle.

Objectif d’un graphique linéaire (Line Chart)
L'objectif principal d'un graphique linéaire est de montrer l'évolution d'une variable continue sur un intervalle de temps donné (chronologie). Il est idéal pour observer des tendances, des accélérations, des baisses ou des cycles (par exemple, les variations de température, le cours d'une action ou la croissance d'une population).

🌟 Exercice 2 : Créer un graphique linéaire pour la variation de température

```python
import matplotlib.pyplot as plt

# Données
jours = ['Lun', 'Mar', 'Mer', 'Jeu', 'Ven', 'Sam', 'Dim']
temperatures = [72, 74, 76, 80, 82, 78, 75]

# Création du graphique linéaire
plt.figure(figsize=(8, 5))
plt.plot(jours, temperatures, marker='o', color='orange', linestyle='-', linewidth=2)

# Personnalisation
plt.title("Évolution de la température sur une semaine", fontsize=14, fontweight='bold')
plt.xlabel("Jour", fontsize=12)
plt.ylabel("Température (°F)", fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)

# Affichage
plt.show()
```

🌟 Exercice 3 : Visualiser les ventes mensuelles avec un graphique à barres

```python
import matplotlib.pyplot as plt

# Données
mois = ['Janvier', 'Février', 'Mars', 'Avril', 'Mai']
ventes = [5000, 5500, 6200, 7000, 7500]

# Création du graphique à barres
plt.figure(figsize=(8, 5))
plt.bar(mois, ventes, color='skyblue', edgecolor='royalblue')

# Personnalisation
plt.title("Montant des ventes mensuelles", fontsize=14, fontweight='bold')
plt.xlabel("Mois", fontsize=12)
plt.ylabel("Montant des ventes ($)", fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Affichage
plt.show()
```

Exercice 4 : Visualiser la distribution du CGPA

```python
import pandas as pd
df = pd.read_csv("Student Mental health.csv", index_col= 0)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Configuration du style Seaborn
sns.set_theme(style="whitegrid")

# Création de l'histogramme
plt.figure(figsize=(8, 5))
sns.histplot(data =df, x='What is your CGPA?', color='purple', shrink=0.8)

# Personnalisation
plt.title("Distribution du CGPA des étudiants", fontsize=14, fontweight='bold')
plt.xlabel("Plage de CGPA", fontsize=12)
plt.ylabel("Nombre d'étudiants", fontsize=12)

# Affichage
plt.show()
```

Exercice 5 : Comparer les niveaux d’anxiété entre différents sexes

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 5))

#Pour comparer des proportions ou des comptes de variables catégorielles avec Seaborn, le paramètre hue est idéal.

# Création du graphique de comparaison (Countplot est le plus adapté ici pour les proportions/comptages)
sns.countplot(data=df, x='Choose your gender', hue='Do you have Anxiety?', palette='pastel')

# Personnalisation
plt.title("Comparaison des niveaux d'anxiété selon le sexe", fontsize=14, fontweight='bold')
plt.xlabel("Genre", fontsize=12)
plt.ylabel("Nombre d'étudiants", fontsize=12)
plt.legend(title="Souffre d'anxiété ?")

# Affichage
plt.show()
```

Exercice 6 : Explorer la relation entre l’âge et les crises de panique

```python
import matplotlib.pyplot as plt
import seaborn as sns

# 0. On règle le problème des index dupliqués en les réinitialisant et en entament un nouvel ordre
df = df.reset_index(drop=True)

# 1. Conversion des réponses textuelles en valeurs numériques (Oui=1, Non=0)
# On utilise map() pour transformer la colonne
df['Panic_Attack_Numeric'] = df['Do you have Panic attack?'].map({'Yes': 1, 'No': 0})

# 2. Création du graphique
plt.figure(figsize=(9, 5))

sns.scatterplot(
    data=df, 
    x='Age', 
    y='Panic_Attack_Numeric', 
    alpha=0.6, 
    s=100, 
    color='crimson', 
    marker='X'
)

# Personnalisation
plt.title("Relation entre l'âge et la survenue des crises de panique", fontsize=14, fontweight='bold')
plt.xlabel("Âge des étudiants", fontsize=12)
plt.ylabel("Crises de panique (1 = Oui, 0 = Non)", fontsize=12)
plt.yticks([0, 1], ['Non (0)', 'Oui (1)']) # Force l'affichage propre des deux choix sur l'axe Y

# Affichage
plt.show()
```
