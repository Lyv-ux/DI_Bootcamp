```python
import numpy as np
```

🌟 Exercice 1 : Création d'un tableau 1D

```python
arr = np.arange(10)

print(arr)
```

🌟 Exercice 2 : Conversion de type


```python
arr = np.array([3.14, 2.17, 0, 1, 2])

arr = arr.astype(int)

print(arr)
```

🌟 Exercice 3 : Tableau multidimensionnel 3×3

```python
arr = np.arange(1, 10).reshape(3, 3)

print(arr)
```

🌟 Exercice 4 : Tableau aléatoire 4×5

```python
arr = np.random.rand(4, 5)

print(arr)
```

🌟 Exercice 5 : Sélectionner la deuxième ligne

```python
arr = np.array([
    [21,22,23,22,22],
    [20,21,22,23,24],
    [21,22,23,22,22]
])

second_row = arr[1]

print(second_row)
```

🌟 Exercice 6 : Inverser l'ordre d'un tableau

```python
arr = np.arange(10)

reversed_arr = arr[::-1]

print(reversed_arr)
```

🌟 Exercice 7 : Matrice identité 4×4

```python
identity = np.eye(4)

print(identity)
```

🌟 Exercice 8 : Somme et moyenne

```python
arr = np.arange(10)

print("Sum:", np.sum(arr))
print("Average:", np.mean(arr))
```

🌟 Exercice 9 : Reshape en matrice 4×5

```python
arr = np.arange(1, 21)

matrix = arr.reshape(4, 5)

print(matrix)
```

🌟 Exercice 10 : Sélection conditionnelle

```python
arr = np.arange(10)

odd_numbers = arr[arr % 2 != 0]

print(odd_numbers)
```
