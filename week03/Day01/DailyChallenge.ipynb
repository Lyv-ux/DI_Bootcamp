```python
# ============================================================
# DAILY CHALLENGE: NUMPY, PANDAS AND MATPLOTLIB INTEGRATION
# ============================================================

# ------------------------------------------------------------
# 1. IMPORTER LES BIBLIOTHÈQUES
# ------------------------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# 2. GÉNÉRER UN JEU DE DONNÉES SYNTHÉTIQUE AVEC NUMPY
# ------------------------------------------------------------
# Températures moyennes mensuelles pour 10 villes sur 12 mois
# Valeurs comprises entre -5°C et 35°C

np.random.seed(42)  # Pour obtenir les mêmes résultats à chaque exécution

temperatures = np.random.uniform(
    low=-5,
    high=35,
    size=(10, 12)
)

# ------------------------------------------------------------
# 3. CRÉER LES NOMS DES VILLES ET DES MOIS
# ------------------------------------------------------------
cities = [
    'New York', 'London', 'Paris', 'Tokyo', 'Sydney',
    'Toronto', 'Berlin', 'Dubai', 'Rome', 'Madrid'
]

months = [
    'Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun',
    'Jul', 'Aug', 'Sep', 'Oct', 'Nov', 'Dec'
]

# ------------------------------------------------------------
# 4. CONVERTIR LE TABLEAU NUMPY EN DATAFRAME PANDAS
# ------------------------------------------------------------
df = pd.DataFrame(
    temperatures,
    index=cities,
    columns=months
)

print("Monthly Temperature Dataset")
print(df)

# ------------------------------------------------------------
# 5. CALCULER LA TEMPÉRATURE MOYENNE ANNUELLE DE CHAQUE VILLE
# ------------------------------------------------------------
df["Annual Average"] = df.mean(axis=1)

print("\nAnnual Average Temperature")
print(df["Annual Average"])

# ------------------------------------------------------------
# 6. IDENTIFIER LA VILLE LA PLUS CHAUDE ET LA PLUS FROIDE
# ------------------------------------------------------------
hottest_city = df["Annual Average"].idxmax()
coldest_city = df["Annual Average"].idxmin()

print("\nCity with Highest Average Temperature:")
print(hottest_city)

print("\nCity with Lowest Average Temperature:")
print(coldest_city)

# ------------------------------------------------------------
# 7. VISUALISATION DES TEMPÉRATURES MOYENNES ANNUELLES
# ------------------------------------------------------------
plt.figure(figsize=(10,6))

plt.bar(
    df.index,
    df["Annual Average"],
    color="skyblue"
)

plt.title("Average Annual Temperature by City")
plt.xlabel("City")
plt.ylabel("Temperature (°C)")
plt.xticks(rotation=45)

plt.show()

# ------------------------------------------------------------
# 8. VISUALISATION DES TENDANCES MENSUELLES
# ------------------------------------------------------------
plt.figure(figsize=(12,6))

for city in cities:
    plt.plot(
        months,
        df.loc[city, months],
        marker="o",
        label=city
    )

plt.title("Monthly Temperature Trends by City")
plt.xlabel("Month")
plt.ylabel("Temperature (°C)")
plt.legend(bbox_to_anchor=(1.05, 1))
plt.grid(True)

plt.show()

# ------------------------------------------------------------
# 9. RAPPORT ET OBSERVATIONS
# ------------------------------------------------------------
print("\n================ SUMMARY REPORT ================\n")

print(
    f"The city with the highest annual average temperature is "
    f"{hottest_city} ({df.loc[hottest_city, 'Annual Average']:.2f} °C)."
)

print(
    f"The city with the lowest annual average temperature is "
    f"{coldest_city} ({df.loc[coldest_city, 'Annual Average']:.2f} °C)."
)

print("\nObservations:")
print("- Temperature patterns vary across cities.")
print("- Some cities remain warm throughout the year.")
print("- Others show larger fluctuations between months.")
print("- The line chart helps compare monthly trends.")
print("- The bar chart makes average temperatures easy to compare.")
```
