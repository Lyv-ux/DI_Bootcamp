```python
# ============================================================
# NUMPY, PANDAS, MATPLOTLIB - DAILY CHALLENGE
# ============================================================

# ------------------------------------------------------------
# IMPORTATION DES BIBLIOTHÈQUES
# ------------------------------------------------------------
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ============================================================
# EXERCICE 1 : MATRIX OPERATIONS
# ============================================================

# Création d'une matrice 3x3
matrix = np.array([
    [2, 1, 3],
    [1, 0, 2],
    [4, 1, 8]
])

# Calcul du déterminant
determinant = np.linalg.det(matrix)

# Calcul de l'inverse
inverse = np.linalg.inv(matrix)

print("EXERCISE 1")
print("Matrix:\n", matrix)
print("\nDeterminant:", determinant)
print("\nInverse:\n", inverse)

# ============================================================
# EXERCICE 2 : STATISTICAL ANALYSIS
# ============================================================

# Génération de 50 nombres aléatoires
data = np.random.rand(50)

# Calcul des statistiques
mean = np.mean(data)
median = np.median(data)
std = np.std(data)

print("\nEXERCISE 2")
print("Mean:", mean)
print("Median:", median)
print("Standard Deviation:", std)

# ============================================================
# EXERCICE 3 : DATE MANIPULATION
# ============================================================

# Création des dates de janvier 2023
dates = np.arange(
    np.datetime64('2023-01-01'),
    np.datetime64('2023-02-01')
)

# Changement de format
formatted_dates = np.datetime_as_string(dates, unit='D')

formatted_dates = np.char.replace(formatted_dates, '-', '/')

print("\nEXERCISE 3")
print(formatted_dates[:10])

# ============================================================
# EXERCICE 4 : NUMPY + PANDAS
# ============================================================

# Création d'un DataFrame aléatoire
df = pd.DataFrame(
    np.random.randint(1, 101, size=(5, 4)),
    columns=['A', 'B', 'C', 'D']
)

# Sélection conditionnelle
selected = df[df['A'] > 50]

# Agrégations
total_sum = df.sum()
average = df.mean()

print("\nEXERCISE 4")
print("\nDataFrame:")
print(df)

print("\nRows where A > 50:")
print(selected)

print("\nSum:")
print(total_sum)

print("\nAverage:")
print(average)

# ============================================================
# EXERCICE 5 : IMAGE REPRESENTATION
# ============================================================

# Création d'une image 5x5 en niveaux de gris
image = np.array([
    [0, 50, 100, 150, 200],
    [25, 75, 125, 175, 225],
    [50, 100, 150, 200, 255],
    [25, 75, 125, 175, 225],
    [0, 50, 100, 150, 200]
])

print("\nEXERCISE 5")
print("Grayscale image represented as NumPy array:")
print(image)

# Affichage de l'image
plt.figure(figsize=(3,3))
plt.imshow(image, cmap='gray')
plt.title("5x5 Grayscale Image")
plt.axis('off')
plt.show()

# ============================================================
# EXERCICE 6 : BASIC HYPOTHESIS TESTING
# ============================================================

# Données fournies
productivity_before = np.random.normal(loc=50, scale=10, size=30)
productivity_after = productivity_before + np.random.normal(loc=5, scale=3, size=30)

# Hypothèse
# H0 : le programme n'améliore pas la productivité
# H1 : le programme améliore la productivité

before_mean = np.mean(productivity_before)
after_mean = np.mean(productivity_after)

difference = after_mean - before_mean

print("\nEXERCISE 6")
print("Average Before Training:", before_mean)
print("Average After Training:", after_mean)
print("Difference:", difference)

if difference > 0:
    print("Conclusion: Training appears to improve productivity.")
else:
    print("Conclusion: No improvement observed.")

# ============================================================
# EXERCICE 7 : COMPLEX ARRAY COMPARISON
# ============================================================

arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([5, 25, 25, 45, 40])

comparison = arr1 > arr2

print("\nEXERCISE 7")
print(comparison)

# ============================================================
# EXERCICE 8 : TIME SERIES MANIPULATION
# ============================================================

# Création de toutes les dates de 2023
time_series = np.arange(
    np.datetime64('2023-01-01'),
    np.datetime64('2024-01-01')
)

# Découpage par trimestre
jan_mar = time_series[:90]
apr_jun = time_series[90:181]
jul_sep = time_series[181:273]
oct_dec = time_series[273:]

print("\nEXERCISE 8")
print("January-March:", jan_mar[:3], "...")
print("April-June:", apr_jun[:3], "...")
print("July-September:", jul_sep[:3], "...")
print("October-December:", oct_dec[:3], "...")

# ============================================================
# EXERCICE 9 : DATA CONVERSION
# ============================================================

# NumPy -> DataFrame
numpy_array = np.array([[1,2],[3,4],[5,6]])

dataframe = pd.DataFrame(
    numpy_array,
    columns=['Col1', 'Col2']
)

# DataFrame -> NumPy
back_to_numpy = dataframe.to_numpy()

print("\nEXERCISE 9")
print("\nDataFrame:")
print(dataframe)

print("\nBack to NumPy:")
print(back_to_numpy)

# ============================================================
# EXERCICE 10 : BASIC VISUALIZATION
# ============================================================

# Création de données aléatoires
random_data = np.random.randint(0, 100, 20)

# Graphique
plt.figure(figsize=(8,4))

plt.plot(
    random_data,
    marker='o'
)

plt.title("Random Numbers Line Graph")
plt.xlabel("Index")
plt.ylabel("Value")
plt.grid(True)

plt.show()

print("\nEXERCISE 10 COMPLETED")
```
