```python
# ==========================================================
# Daily Challenge: Preprocess & Fine-Tune Transformer Models
# Tasks 1 → 5 in one cell
# ==========================================================

# =========================
# Imports
# =========================

import pandas as pd
from transformers import BertTokenizer, XLMRobertaTokenizer
from sklearn.model_selection import StratifiedKFold

# =========================
# 1. Load Tokenizers
# =========================

bert_tokenizer = BertTokenizer.from_pretrained("bert-base-uncased")
xlmr_tokenizer = XLMRobertaTokenizer.from_pretrained("xlm-roberta-base")

# =========================
# 2. Tokenizing Text
# =========================

sentence1 = "Transformers are powerful models for NLP."
sentence2 = "They can be fine-tuned for classification tasks."

# BERT tokenization (two sentences)
bert_encoding = bert_tokenizer.encode_plus(
    sentence1,
    sentence2,
    add_special_tokens=True,
    max_length=32,
    padding="max_length",
    truncation=True,
    return_attention_mask=True,
    return_tensors="pt"
)

print("===== BERT TOKENIZATION =====")
print("Input IDs:")
print(bert_encoding["input_ids"])

print("\nAttention Mask:")
print(bert_encoding["attention_mask"])

print("\nDecoded Text:")
print(
    bert_tokenizer.decode(
        bert_encoding["input_ids"][0],
        skip_special_tokens=False
    )
)

# XLM-RoBERTa tokenization
xlmr_encoding = xlmr_tokenizer.encode_plus(
    sentence1,
    sentence2,
    add_special_tokens=True,
    max_length=32,
    padding="max_length",
    truncation=True,
    return_attention_mask=True,
    return_tensors="pt"
)

print("\n===== XLM-RoBERTA TOKENIZATION =====")
print("Input IDs:")
print(xlmr_encoding["input_ids"])

print("\nAttention Mask:")
print(xlmr_encoding["attention_mask"])

print("\nDecoded Text:")
print(
    xlmr_tokenizer.decode(
        xlmr_encoding["input_ids"][0],
        skip_special_tokens=False
    )
)

# =========================
# 3. Preparing Input Data
# =========================

print("\n===== SPECIAL TOKENS =====")
print(xlmr_tokenizer.special_tokens_map)

print("\nVocabulary Size:")
print(xlmr_tokenizer.vocab_size)

sample_text = "This is a sample sentence."

prepared_input = xlmr_tokenizer.encode_plus(
    sample_text,
    padding="max_length",
    truncation=True,
    max_length=20,
    return_attention_mask=True
)

print("\nPrepared Input:")
print(prepared_input)

# =========================
# 4. Load Dataset
# =========================

# Replace file names with your dataset paths
train_df = pd.read_csv("train.csv")
test_df = pd.read_csv("test.csv")

print("\n===== TRAIN DATA =====")
print(train_df.head())

print("\nTrain Shape:")
print(train_df.shape)

print("\n===== TEST DATA =====")
print(test_df.head())

print("\nTest Shape:")
print(test_df.shape)

# Assume:
# text column = "text"
# target column = "label"
# Change names if your dataset uses different columns

TEXT_COLUMN = "text"
LABEL_COLUMN = "label"

print("\nAvailable Columns:")
print(train_df.columns)

# =========================
# 5. Stratified K-Fold
# =========================

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

train_folds = []
valid_folds = []

for fold, (train_idx, valid_idx) in enumerate(
    skf.split(
        train_df[TEXT_COLUMN],
        train_df[LABEL_COLUMN]
    )
):

    train_split = train_df.iloc[train_idx]
    valid_split = train_df.iloc[valid_idx]

    train_folds.append(train_split)
    valid_folds.append(valid_split)

    print(f"\n===== FOLD {fold+1} =====")
    print("Train Shape:", train_split.shape)
    print("Validation Shape:", valid_split.shape)

print("\nTotal Folds Created:", len(train_folds))
```
