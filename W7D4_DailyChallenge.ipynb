1. Understanding LLM Evaluation
Why is evaluating LLMs more complex than traditional software?
Traditional software usually gives the same output for the same input. Its correctness can be checked with clear tests.
LLMs are different because they generate text and can produce different answers for the same question. Their responses must be evaluated for accuracy, relevance, fluency, reasoning, safety, and bias. This makes evaluation more difficult.
Why is it important to evaluate an LLM's safety?
Safety evaluation is important because LLMs may:

Generate false information.
Produce harmful or offensive content.
Reinforce social biases.
Leak sensitive information.
Give dangerous advice.

Testing safety helps ensure that users receive responsible and trustworthy responses.
How does adversarial testing help improve LLMs?
Adversarial testing uses difficult or misleading prompts to find weaknesses in a model.
It helps developers:

Detect vulnerabilities.
Identify hallucinations.
Find bias problems.
Improve robustness.
Make the model safer.

What are the limitations of automated metrics compared to human evaluation?
Automated metrics are fast and inexpensive, but they only compare text patterns.
Limitations:

They may ignore meaning.
They cannot fully judge creativity.
They may not capture context.
They do not always reflect human preferences.

Human evaluation is usually more accurate because people can assess meaning, fluency, coherence, and usefulness.

Limitations of BLEU and ROUGE
BLEU and ROUGE have several limitations:

They rely on word overlap.
They do not understand synonyms well.
They do not measure factual correctness.
They cannot evaluate creativity effectively.
They may penalize valid alternative wording.

For creative writing, storytelling, and dialogue systems, these metrics may underestimate quality.

Possible Improvements
Alternative evaluation methods include:

BERTScore: compares semantic meaning instead of exact words.
Human Evaluation: evaluates fluency, relevance, and usefulness.
GPT-based Evaluation: uses another LLM as a judge.
Task-specific evaluation: checks factual accuracy and reasoning.

Combining automated and human evaluation usually gives the best results.

6. Comparative Analysis of Evaluation Methods
Chosen NLP Task: Text Summarization
BLEU
Advantages

Easy to calculate.
Widely used.

Disadvantages

Depends on exact words.
Misses semantic similarity.


ROUGE
Advantages

Measures content overlap.
Popular for summarization.

Disadvantages

Does not understand meaning well.
May reward longer summaries.


BERTScore
Advantages

Uses contextual embeddings.
Captures semantic meaning.
Handles synonyms better.

Disadvantages

More computationally expensive.


Human Evaluation
Advantages

Best at measuring quality.
Evaluates usefulness, coherence, and factuality.

Disadvantages

Expensive.
Time-consuming.


Which Metric Is Best?
For text summarization, BERTScore is often the most appropriate automated metric because it evaluates semantic similarity rather than exact word matches.
However, the best overall evaluation approach is a combination of:

ROUGE
BERTScore
Human Evaluation

This combination provides a more complete assessment of summary quality.
