```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Load the dataset
df = pd.read_csv(
    "ex2data1.txt",
    header=None,
    names=["Exam1", "Exam2", "Admitted"]
)

# Display first rows
print(df.head())
```

```python
# Separate admitted and not admitted students
admitted = df[df["Admitted"] == 1]
not_admitted = df[df["Admitted"] == 0]

# Scatter plot
plt.scatter(admitted["Exam1"], admitted["Exam2"], color="green", label="Admitted")
plt.scatter(not_admitted["Exam1"], not_admitted["Exam2"], color="red", label="Not Admitted")

plt.xlabel("Exam 1 Score")
plt.ylabel("Exam 2 Score")
plt.title("Admission Based on Exam Scores")
plt.legend()
plt.show()
```

```python
# Features and target
X = df[["Exam1", "Exam2"]]
y = df["Admitted"]

# Create and train the model
model = LogisticRegression()
model.fit(X, y)
```

```python
# Predictions
y_pred = model.predict(X)

# Accuracy
accuracy = accuracy_score(y, y_pred)
print("Model Accuracy:", accuracy)
```

Model Evaluation and Interpretation

Logistic Regression predicts the probability of admission.
Higher exam scores increase the chance of admission.
Accuracy tells us how well the model performs.
A higher accuracy means a better model.


Conclusion
Logistic Regression is good for binary problems.
It is simple and easy to interpret.
Exam scores are important features for admission prediction.
This model helps understand admission decisions.
