## Exercise 1: Open Source Levels Reflection
**Use:** Fully Open, Weights Released, Architecture Only; model components; fine-tuning/domain adaptation; healthcare compliance.

**Step-by-step**
1) Gather definitions for the three openness levels.  
2) Identify key characteristics (what is open; what you can/cannot do).  
3) Compare side by side in a table (what's open vs. impact).  
4) Write a 3?5 sentence comparative paragraph.  
5) Answer the healthcare prompt (1?2 sentences) about retraining on clinical data.

**Deliverables:** paragraph + healthcare answer.

```python
open_source_levels = {
    "Fully Open": {
        "definition": "All components (code, weights, data, architecture) are publicly available, often under a permissive license.",
        "what_is_open": ["Training code", "Model weights", "Training data", "Model architecture", "Evaluation metrics"],
        "what_you_can_do": ["Inspect, modify, and redistribute all components", "Reproduce training and fine-tuning", "Use for commercial purposes (usually)"],
        "what_you_cannot_do": ["None, typically (depending on license restrictions)"],
    },
    "Weights Released": {
        "definition": "Model weights are publicly available, but training code or data may be proprietary.",
        "what_is_open": ["Model weights", "Model architecture"],
        "what_you_can_do": ["Run inference with the model", "Fine-tune the model on custom data", "Inspect the model's structure"],
        "what_you_cannot_do": ["Reproduce original training from scratch (without full code/data)", "Deeply understand training process without proprietary data/code"],
    },
    "Architecture Only": {
        "definition": "Only the model's architecture (e.g., neural network structure) is public, without weights or training data.",
        "what_is_open": ["Model architecture (e.g., paper, code for building the structure)"],
        "what_you_can_do": ["Implement the architecture", "Train a new model from scratch using the published architecture"],
        "what_you_cannot_do": ["Perform inference with a pre-trained model", "Fine-tune a pre-trained model", "Access any pre-trained knowledge"],
    },
}
open_source_levels
```

```python
comparison_table = """| Openness level | What's open? | Impact on retraining/modifying |
| --- | --- | --- |
| Fully Open | Training code, model weights, data, architecture, evaluation metrics | Full inspection, modification, redistribution; easy reproduction and fine-tuning. |
| Weights Released | Model weights, model architecture | Inference and fine-tuning possible; limited understanding of original training without code/data. |
| Architecture Only | Model architecture (structure) | Can implement architecture and train new models from scratch; no pre-trained inference or fine-tuning. |"""
print(comparison_table)
```

```python
comparative_paragraph = """"Fully Open" LLMs provide complete access to all components, allowing for deep inspection, modification, and reproduction of the entire training process. In contrast, "Weights Released" models offer pre-trained weights and architecture, enabling inference and fine-tuning but restricting full understanding or reproduction of the original training due to proprietary code or data. Lastly, "Architecture Only" models expose only the structural design, requiring users to train new models from scratch if they wish to utilize the architecture, as no pre-trained weights or data are provided. The degree of openness significantly impacts what developers can do, from full customization to only architectural implementation."""
print(comparative_paragraph)
```

```python
healthcare_prompt_answer = """For healthcare compliance and ethical use with clinical data, "Fully Open" LLMs are ideal due to complete transparency, auditability, and control over all components. While "Weights Released" models allow fine-tuning, the proprietary nature of their initial training data and code can pose significant challenges for regulatory approval and understanding potential biases when adapted for sensitive healthcare applications."""
print(healthcare_prompt_answer)
```
